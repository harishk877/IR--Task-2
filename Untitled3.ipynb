## 1. Problem Statement

## 2. Environment Setup & Data Loading
First, we will import the required Python libraries. We'll use the `datasets` library from Hugging Face to load the official IMDb dataset efficiently and select a balanced sample of 10,000 reviews (5,000 positive, 5,000 negative) to keep the execution times highly manageable while maintaining statistical validity.

```python
import re
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
from datasets import load_dataset

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import LinearSVC
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

print("Libraries successfully imported!")
```

```python
# Load the dataset using a reliable direct CSV mirror
print("Loading IMDb dataset from CSV mirror...")
url = "https://raw.githubusercontent.com/Ankit152/IMDB-Sentiment-Analysis/master/IMDB-Dataset.csv"
df_all = pd.read_csv(url)

# Map positive/negative to 1/0
df_all['label'] = df_all['sentiment'].map({'positive': 1, 'negative': 0})
df_all = df_all.rename(columns={'review': 'text'})

# Create a balanced subset of 10,000 samples (5,000 Positive, 5,000 Negative)
df_pos = df_all[df_all['label'] == 1].sample(n=5000, random_state=42)
df_neg = df_all[df_all['label'] == 0].sample(n=5000, random_state=42)

df = pd.concat([df_pos, df_neg]).sample(frac=1, random_state=42).reset_index(drop=True)
print(f"Subset successfully created. Shape: {df.shape}")
```

## 3. Dataset Exploration
We explore the basic properties of the dataset such as checking for missing values, duplicated entries, and visualizing the target label distributions to ensure our subset is balanced.

```python
# Display basic info and check for null values or duplicates
print("--- Dataset Information ---")
print(df.info())

print("\n--- Null Value Count ---")
print(df.isnull().sum())

print("\n--- Duplicated Rows Count ---")
print(df.duplicated(subset=['text']).sum())

# Display sample reviews
print("\n--- Sample Reviews ---")
display(df.head(3))
```

## 4. Text Preprocessing
Text preprocessing cleans noise from raw text. We will implement a function to:
1. Convert text to **lowercase**.
2. Remove **HTML tags** (like `<br />`).
3. Remove **URLs**.
4. Remove **special characters** and extra white spaces, while keeping important negations ("not", "no", "never") which contain critical sentiment information.

```python
def clean_text(text):
    # 1. Lowercase
    text = text.lower()

    # 2. Remove HTML Tags
    text = re.sub(r'<[^>]+>', ' ', text)

    # 3. Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', ' ', text)

    # 4. Remove special characters but keep spaces and alphanumeric words
    # Also preserving common contraction forms for negations
    text = re.sub(r"[^a-zA-Z0-9\s']", ' ', text)

    # 5. Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()

    return text

# Apply the text cleaning function
df['cleaned_text'] = df['text'].apply(clean_text)

# Demo preprocessing
print("Original:")
print(df['text'].iloc[0][:150] + "...")
print("\nCleaned:")
print(df['cleaned_text'].iloc[0][:150] + "...")
```

## 5. Train-Test Split & TF-IDF Feature Extraction
We split the dataset into 80% training and 20% test data. We use stratification to ensure class proportions are identical in both sets.
To avoid data leakage, we fit the `TfidfVectorizer` **only** on the training set and transform the test set using that fitted vectorizer.

```python
X = df['cleaned_text']
y = df['label']

X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print(f"Training instances: {X_train_raw.shape[0]}")
print(f"Testing instances: {X_test_raw.shape[0]}")
```

```python
# Feature Extraction using TF-IDF
vectorizer = TfidfVectorizer(max_features=5000, ngram_range=(1, 2), min_df=2)

# Fit on Train only and Transform both
X_train = vectorizer.fit_transform(X_train_raw)
X_test = vectorizer.transform(X_test_raw)

print(f"TF-IDF Matrix Shape (Train): {X_train.shape}")
print(f"TF-IDF Matrix Shape (Test): {X_test.shape}")
```

## 6. Model Implementations
Now we will initialize and train our 4 classification algorithms using the exact same preprocessed datasets.

```python
# 1. Decision Tree
dt_model = DecisionTreeClassifier(max_depth=30, random_state=42)
dt_model.fit(X_train, y_train)
dt_preds = dt_model.predict(X_test)

# 2. KNN
knn_model = KNeighborsClassifier(n_neighbors=5, metric='cosine', algorithm='brute')
knn_model.fit(X_train, y_train)
knn_preds = knn_model.predict(X_test)

# 3. Support Vector Machine (Linear SVC)
svm_model = LinearSVC(random_state=42, max_iter=2000)
svm_model.fit(X_train, y_train)
svm_preds = svm_model.predict(X_test)

# 4. Multinomial Naive Bayes
nb_model = MultinomialNB(alpha=1.0)
nb_model.fit(X_train, y_train)
nb_preds = nb_model.predict(X_test)

print("All 4 models trained successfully!")
```

## 7. Model Evaluation and Performance Comparison
We collect Accuracy, Precision, Recall, and F1-score for each model and compile them into a unified, formatted metrics table.

```python
models_preds = {
    'Decision Tree': dt_preds,
    'KNN (K=5)': knn_preds,
    'Linear SVM': svm_preds,
    'Naive Bayes': nb_preds
}

results_list = []
for name, preds in models_preds.items():
    acc = accuracy_score(y_test, preds)
    prec = precision_score(y_test, preds)
    rec = recall_score(y_test, preds)
    f1 = f1_score(y_test, preds)
    results_list.append({
        'Model': name,
        'Accuracy': round(acc, 4),
        'Precision': round(prec, 4),
        'Recall': round(rec, 4),
        'F1-Score': round(f1, 4)
    })

df_results = pd.DataFrame(results_list)
display(df_results)

# Save comparison table to CSV
df_results.to_csv('model_comparison_results.csv', index=False)
print("Saved comparison results to model_comparison_results.csv")
```

## 8. Visualizations
Let's generate publication-ready visualizations:
1. A grouped bar chart comparing all 4 metrics across the classifiers.
2. A comparison chart focused entirely on the F1-score.
3. Confusion matrices for all classifiers arranged in a clean 2x2 grid layout.

```python
# 1. Grouped Bar Chart
df_melted = df_results.melt(id_vars='Model', var_name='Metric', value_name='Score')

plt.figure(figsize=(12, 6))
sns.barplot(data=df_melted, x='Model', y='Score', hue='Metric', palette='viridis')
plt.title('Classifier Performance Comparison', fontsize=14, fontweight='bold')
plt.ylabel('Score (0 to 1)')
plt.ylim(0.5, 1.0)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('performance_comparison_chart.png', dpi=300)
plt.show()
```

```python
# 2. Separate F1-Score Chart
plt.figure(figsize=(8, 5))
sns.barplot(data=df_results, x='Model', y='F1-Score', palette='magma')
plt.title('F1-Score Comparison across Models', fontsize=12, fontweight='bold')
plt.ylabel('F1-Score')
plt.ylim(0.5, 1.0)
for index, row in df_results.iterrows():
    plt.text(index, row['F1-Score'] + 0.01, f"{row['F1-Score']:.4f}", ha='center', va='bottom', fontsize=10)
plt.tight_layout()
plt.savefig('f1_score_comparison.png', dpi=300)
plt.show()
```

```python
# 3. Confusion Matrices 2x2 Subplot
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.ravel()

for idx, (name, preds) in enumerate(models_preds.items()):
    cm = confusion_matrix(y_test, preds)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[idx], cbar=False,
                xticklabels=['Negative', 'Positive'], yticklabels=['Negative', 'Positive'])
    axes[idx].set_title(f'{name} Confusion Matrix', fontsize=12, fontweight='bold')
    axes[idx].set_xlabel('Predicted Class')
    axes[idx].set_ylabel('True Class')

plt.tight_layout()
plt.savefig('confusion_matrices_grid.png', dpi=300)
plt.show()
```

## 9. Confusion Matrix Analysis
Here we analyze the confusion matrices programmatically, explaining the counts of True Positives, True Negatives, False Positives, and False Negatives for each of our models based on actual outputs.

```python
for name, preds in models_preds.items():
    tn, fp, fn, tp = confusion_matrix(y_test, preds).ravel()
    print(f"=== {name} Analysis ===")
    print(f"- True Negatives (TN - Correctly predicted negative reviews): {tn}")
    print(f"- False Positives (FP - Negative reviews classified as positive): {fp}")
    print(f"- False Negatives (FN - Positive reviews classified as negative): {fn}")
    print(f"- True Positives (TP - Correctly predicted positive reviews): {tp}")
    total_errors = fp + fn
    print(f"- Total Misclassifications (FP + FN): {total_errors}\n")
```

## 10. What-If Experiment (KNN Neighbors Sensitivity Analysis)
We run a sensitivity study evaluating how performance varies with different $K$ parameters ($K = 3$, $K = 5$, and $K = 7$) on the same TF-IDF features.

```python
k_values = [3, 5, 7]
what_if_results = []

for k in k_values:
    knn_temp = KNeighborsClassifier(n_neighbors=k, metric='cosine', algorithm='brute')
    knn_temp.fit(X_train, y_train)
    temp_preds = knn_temp.predict(X_test)

    acc = accuracy_score(y_test, temp_preds)
    prec = precision_score(y_test, temp_preds)
    rec = recall_score(y_test, temp_preds)
    f1 = f1_score(y_test, temp_preds)

    what_if_results.append({
        'K-Value': f'K={k}',
        'Accuracy': round(acc, 4),
        'Precision': round(prec, 4),
        'Recall': round(rec, 4),
        'F1-Score': round(f1, 4)
    })

df_whatif = pd.DataFrame(what_if_results)
display(df_whatif)

# Save What-if results
df_whatif.to_csv('knn_sensitivity_results.csv', index=False)

# Visualize What-if Sensitivity Analysis
plt.figure(figsize=(8, 5))
df_whatif_melted = df_whatif.melt(id_vars='K-Value', var_name='Metric', value_name='Score')
sns.lineplot(data=df_whatif_melted, x='K-Value', y='Score', hue='Metric', marker='o', linewidth=2.5)
plt.title('KNN Performance Sensitivity across different K values', fontweight='bold')
plt.ylabel('Score')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('knn_sensitivity_plot.png', dpi=300)
plt.show()
```

## 11. Best Model Identification & Interactive Inference Pipeline
Below, we automatically identify the model that scored the highest F1-score and create an inference pipeline that allows you to submit custom text snippets to test each of the models live.

```python
# Auto-identify the best performing algorithm
best_model_row = df_results.loc[df_results['F1-Score'].idxmax()]
print("=== BEST PERFORMING CLASSIFIER ===")
print(best_model_row)
```

```python
# Interactive Prediction Demo
def predict_sentiment(review_text):
    cleaned = clean_text(review_text)
    vec_text = vectorizer.transform([cleaned])

    predictions = {
        'Decision Tree': dt_model.predict(vec_text)[0],
        'KNN (K=5)': knn_model.predict(vec_text)[0],
        'Linear SVM': svm_model.predict(vec_text)[0],
        'Naive Bayes': nb_model.predict(vec_text)[0]
    }

    print(f"Review: \"{review_text}\"\n")
    for model, pred in predictions.items():
        label_str = "Positive" if pred == 1 else "Negative"
        print(f"- {model:15s} Predicted: {label_str}")
    print("-" * 50)

# Demo 4 Sample Reviews
samples = [
    "This was an absolute masterpiece! Brilliant acting and breathtaking storytelling.",
    "Waste of time. The movie was boring, predictable, and had a terrible ending.",
    "Not bad, but not particularly good either. It lacked depth.",
    "I was pleasantly surprised! It was not a disappointment."
]

for sample in samples:
    predict_sentiment(sample)
```
